```python
#dicionarios com python
# 1 - todo dicionario é uma data frame (df)

turmaad6 = {
    # chave é unitario
    # valor por não ser unitario
    "professores" : ["Felipe", "Luyz", "Priscila"],
    "alunos": ["André", "Paulo", "Aline", "Paola"]

}
print(turmaad6['professores'][1])
```

```python
# vamos criar as funçõoes
# ver os dados
# adicionar novo professor
# adicionar novo aluno
# remover aluno
# remover professor
def ver_dados():
  print('----professores cadastrados----')
  for i in turmaad6['professores']:
    print(i)

    print('--- alunos cadastrados ---')
    for i in turmaad6['alunos']:
      print(i)

def add_prof():
  novo_professor = input('Qual o nome do professor: ')
  # chama a lista. append(dado)
  turmaad6['professores'].append(novo_professor)
  print(f'o professor {novo_professor} foi adicionado')
  ver_dados()

def add_aluno():
  novo_aluno = input("Qual o nome do aluno: ")
  turmaad6["alunos"].append(novo_aluno)
  print(f"O aluno {novo_aluno} foi adicionado")
  ver_dados()

def remover_aluno():
  # se eu quero escolher quem eu vou remover
  # eu vou usar o pop(posicao)
  posicao = 1
  for i in turmaad6['alunos']:
    print(f'{posicao} - {i}')
    posicao += 1
  remover = int(input('informe a posição do aluno'))
  # para remover
  turmaad6['alunos'].pop(remover -1)
  ver_dados()

def remover_professor():
  posicao = 1
  for i in turmaad6['professores']:
    print(f"{posicao} - {i}")
    posicao +=1

  remover = int(input('Informe a posição do professor'))
  turmaad6['professores'].pop(remover -1)
  ver_dados()
```

```python
remover_professor()
```

```python
remover_aluno()
```

```python
add_prof()
add_aluno()
```
